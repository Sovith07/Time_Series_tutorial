# Smoothing in Time Series

## 1. What is Smoothing?

**Smoothing** is a technique used in time-series analysis to reduce short-term fluctuations or random noise and reveal the underlying pattern of the data.

A time series can be thought of as:

$$
\boxed{
\text{Observed Data}
====================

\text{Underlying Pattern}
+
\text{Noise}
}
$$

The underlying pattern may contain:

* **Level**
* **Trend**
* **Seasonality**
* **Cyclical patterns**

Smoothing attempts to reduce the effect of irregular fluctuations so that these components become easier to identify.

---

## 2. Why Do We Need Smoothing?

Consider the following sales data:

```text
100 → 130 → 95 → 145 → 120 → 160 → 135
```

There are large short-term fluctuations, making the underlying trend difficult to identify.

After smoothing:

```text
108 → 112 → 120 → 125 → 135 → 140
```

the underlying upward movement becomes easier to observe.

Therefore:

$$
\boxed{
\text{Smoothing}
\rightarrow
\text{Reduce Noise}
\rightarrow
\text{Identify Underlying Pattern}
}
$$

---

## 3. Main Objectives of Smoothing

Smoothing is mainly used to:

1. Reduce random fluctuations
2. Identify the underlying trend
3. Visualize time-series patterns
4. Estimate the current level of a series
5. Generate forecasts
6. Separate signal from noise
7. Make trend and seasonality easier to analyze

---

# 4. Main Smoothing Techniques

The important smoothing techniques are:

```text
Smoothing
│
├── Moving Average
│   │
│   ├── Simple Moving Average
│   ├── Trailing Moving Average
│   └── Weighted Moving Average
│
└── Exponential Smoothing
    │
    ├── Simple Exponential Smoothing
    ├── Holt's Method
    └── Holt-Winters Method
```

# 5. Moving Average

A **Moving Average (MA)** calculates the average of a fixed number of consecutive observations.

For a $k$-period trailing moving average:

$$
\boxed{
MA_t =
\frac{1}{k}
\sum_{i=0}^{k-1} Y_{t-i}
}
$$

where:

* $Y_t$ = observation at time $t$
* $k$ = window size
* $MA_t$ = moving average

A moving average smooths the series by averaging nearby observations.

---

# 6. Example of Moving Average

Suppose we have:

| Month | Sales |
| ----- | ----: |
| Jan   |   100 |
| Feb   |   120 |
| Mar   |   140 |
| Apr   |   160 |
| May   |   180 |

For a **3-period moving average**:

### March

$$
MA_{\text{Mar}}
===============

\frac{100+120+140}{3}
$$

$$
\boxed{
MA_{\text{Mar}}=120
}
$$

### April

$$
MA_{\text{Apr}}
===============

\frac{120+140+160}{3}
$$

$$
\boxed{
MA_{\text{Apr}}=140
}
$$

### May

$$
MA_{\text{May}}
===============

\frac{140+160+180}{3}
$$

$$
\boxed{
MA_{\text{May}}=160
}
$$

---

# 7. Effect of Window Size

The number of observations used in the moving average is called the **window size**.

### 3-period MA

$$
MA_t =
\frac{Y_t+Y_{t-1}+Y_{t-2}}{3}
$$

### 5-period MA

$$
MA_t =
\frac{
Y_t+Y_{t-1}+Y_{t-2}+Y_{t-3}+Y_{t-4}
}{5}
$$

### 12-period MA

$$
MA_t =
\frac{1}{12}
\sum_{i=0}^{11}Y_{t-i}
$$

A larger window produces stronger smoothing.

```text
Small window
     ↓
Less smoothing
More responsive

Large window
     ↓
More smoothing
Less responsive
```

Therefore:

$$
\boxed{
\text{Larger window}
\rightarrow
\text{More smoothing}
\rightarrow
\text{Less responsiveness}
}
$$

---

# 8. Trailing Moving Average

A **Trailing Moving Average** uses the current observation and previous observations.

It does **not use future observations**.

For a $k$-period trailing moving average:

$$
\boxed{
TMA_t =
\frac{
Y_t+Y_{t-1}+\cdots+Y_{t-k+1}
}{k}
}
$$

For a 3-period trailing MA:

$$
\boxed{
TMA_t =
\frac{
Y_t+Y_{t-1}+Y_{t-2}
}{3}
}
$$

---

# 9. Example of Trailing Moving Average

Suppose:

| Month | Sales |
| ----- | ----: |
| Jan   |   100 |
| Feb   |   120 |
| Mar   |   140 |
| Apr   |   160 |
| May   |   180 |

The 3-month trailing MA is:

### March

$$
TMA_{\text{Mar}}
================

# \frac{100+120+140}{3}

120
$$

### April

$$
TMA_{\text{Apr}}
================

# \frac{120+140+160}{3}

140
$$

### May

$$
TMA_{\text{May}}
================

# \frac{140+160+180}{3}

160
$$

Therefore:

| Month | Actual | 3-Month Trailing MA |
| ----- | -----: | ------------------: |
| Jan   |    100 |                 NaN |
| Feb   |    120 |                 NaN |
| Mar   |    140 |                 120 |
| Apr   |    160 |                 140 |
| May   |    180 |                 160 |

The first $k-1$ observations do not have enough previous observations to calculate the moving average.

---

# 10. Why Is It Called "Trailing"?

The moving window **trails behind the current time**.

For example, at April:

```text
Jan     Feb     Mar     Apr     May
100     120     140     160     180
        └───────────────┘
          3 observations
```

The trailing MA at April uses:

$$
Feb,\ Mar,\ Apr
$$

It does not use May.

Therefore:

$$
\boxed{
\text{Trailing MA uses current + past observations}
}
$$

---

# 11. Trailing MA vs Centered MA

### Trailing Moving Average

$$
\boxed{
TMA_t =
\frac{
Y_t+Y_{t-1}+Y_{t-2}
}{3}
}
$$

Uses:

$$
t,\ t-1,\ t-2
$$

### Centered Moving Average

$$
\boxed{
CMA_t =
\frac{
Y_{t-1}+Y_t+Y_{t+1}
}{3}
}
$$

Uses:

$$
t-1,\ t,\ t+1
$$

The centered MA uses a **future observation**.

Therefore:

$$
\boxed{
\text{Trailing MA}
\rightarrow
\text{Suitable for real-time forecasting}
}
$$

while:

$$
\boxed{
\text{Centered MA}
\rightarrow
\text{Mainly useful for historical analysis and smoothing}
}
$$

---

# 12. Trailing MA for Forecasting

Suppose:

```text
Jan = 100
Feb = 120
Mar = 140
```

The 3-month trailing average is:

$$
\frac{100+120+140}{3}=120
$$

This value can be used as the forecast for the next period:

$$
\boxed{
\hat{Y}_{Apr}=120
}
$$

Thus, a trailing moving average can be used both for:

* Smoothing
* Forecasting

# 13. Naive Forecast

The **Naive Forecast** is one of the simplest forecasting methods.

It assumes:

> The next observation will be equal to the most recent observation.

Therefore:

$$
\boxed{
\hat{Y}_{t+1}=Y_t
}
$$

where:

* $Y_t$ = latest observed value
* $\hat{Y}_{t+1}$ = forecast for the next period

---

# 14. Example of Naive Forecast

Suppose:

| Month | Sales |
| ----- | ----: |
| Jan   |   100 |
| Feb   |   120 |
| Mar   |   140 |

The latest observation is:

$$
Y_{\text{Mar}}=140
$$

Therefore:

$$
\boxed{
\hat{Y}_{\text{Apr}}=140
}
$$

If April actually becomes 150:

$$
\hat{Y}_{\text{Apr}}=140
$$

$$
Y_{\text{Apr}}=150
$$

Forecast error:

$$
e_{\text{Apr}}
==============

Y_{\text{Apr}}-\hat{Y}_{\text{Apr}}
$$

$$
e_{\text{Apr}}=150-140=10
$$

---

# 15. Naive Forecast for Multiple Periods

An important property of the naive forecast is that the forecast remains constant for all future periods.

Suppose:

$$
Y_t=140
$$

Then:

$$
\hat{Y}_{t+1}=140
$$

$$
\hat{Y}_{t+2}=140
$$

$$
\hat{Y}_{t+3}=140
$$

and so on.

```text
Last observed value = 140

Forecast:
t+1 → 140
t+2 → 140
t+3 → 140
t+4 → 140
```

---

# 16. Naive Forecast in Python

Suppose:

```python
series = df_mm["y"]
```

The one-step-ahead naive forecast can be created using:

```python
naive_forecast = series.shift(1)
```

Or added directly to the DataFrame:

```python
df_mm["naive_forecast"] = df_mm["y"].shift(1)
```

This implements:

$$
\boxed{
\hat{Y}*t=Y*{t-1}
}
$$

---

# 17. Naive Forecast for Test Data

Suppose we split the data:

```python
train = series.iloc[:-12]
test = series.iloc[-12:]
```

The last training observation becomes the forecast for the first test observation.

For a multi-step naive baseline:

```python
naive_forecast = [train.iloc[-1]] * len(test)
```

This produces the same forecast for every test observation.

# 18. Seasonal Naive Forecast

The normal naive method uses the **previous observation**.

For seasonal data, we can instead use a **Seasonal Naive Forecast**.

The seasonal naive forecast assumes:

> The value will be similar to the value from the same season in the previous cycle.

The formula is:

$$
\boxed{
\hat{Y}*t=Y*{t-s}
}
$$

where:

* $s$ = seasonal period

---

# 19. Example of Seasonal Naive Forecast

Suppose we have monthly data:

$$
s=12
$$

For January 2026, the forecast is based on January 2025:

$$
\boxed{
\hat{Y}_{\text{Jan 2026}}
=========================

Y_{\text{Jan 2025}}
}
$$

For February:

$$
\boxed{
\hat{Y}_{\text{Feb 2026}}
=========================

Y_{\text{Feb 2025}}
}
$$

and so on.

This is especially useful when strong seasonality exists.

---

# 20. Seasonal Naive in Python

For monthly data:

```python
seasonal_naive = series.shift(12)
```

because:

$$
s=12
$$

For weekly seasonality in daily data:

```python
seasonal_naive = series.shift(7)
```

For quarterly seasonality:

```python
seasonal_naive = series.shift(4)
```

Therefore:

$$
\boxed{
m=s
}
$$

when $m$ is used to represent the seasonal period.

# 23. Moving Average as a Forecast

A trailing moving average can be used as a forecasting method.

For a $k$-period moving average:

$$
\boxed{
\hat{Y}_{t+1}
=============

\frac{1}{k}
\sum_{i=0}^{k-1}Y_{t-i}
}
$$

For a 3-period MA:

$$
\boxed{
\hat{Y}_{t+1}
=============

\frac{
Y_t+Y_{t-1}+Y_{t-2}
}{3}
}
$$

The forecast changes as new observations become available.

---

# 24. Recursive Moving Average Forecasting

Suppose:

```text
Y₁ = 100
Y₂ = 120
Y₃ = 140
```

For a 3-period MA:

$$
\hat{Y}_4
=========

# \frac{100+120+140}{3}

120
$$

After observing $Y_4$, we can calculate:

$$
\hat{Y}_5
=========

\frac{
Y_2+Y_3+Y_4
}{3}
$$

So the moving-average forecast is updated as new actual observations arrive.

### Important distinction

For **multi-step forecasting**, if future actual observations are not available, a moving-average method must use its own previous forecasts or another recursive strategy. Therefore, the simple rolling-average formula is naturally a **one-step-ahead / updating forecast** when actual observations become available sequentially.

# 25. Exponential Smoothing

Another important smoothing method is **Exponential Smoothing**.

The basic formula is:

$$
\boxed{
\hat{Y}_{t+1}=
\alpha Y_t+
(1-\alpha)\hat{Y_t}
}
$$

where:

$$
0<\alpha<1
$$

The recent observation receives weight $\alpha$, while the previous smoothed value receives weight:

$$
1-\alpha
$$

The previous smoothed value itself contains information from earlier observations, so older observations receive progressively smaller weights.

---

# 26. Effect of Alpha

### High $\alpha$

For example:

$$
\alpha=0.8
$$

The model reacts quickly to recent changes.

```text
High α
 ↓
More responsive
Less smoothing
```

### Low $\alpha$

For example:

$$
\alpha=0.1
$$

The model changes slowly.

```text
Low α
 ↓
More smoothing
More stable
```

Therefore:

$$
\boxed{
\alpha\uparrow
\Rightarrow
\text{Greater importance to recent observations}
}
$$

and:

$$
\boxed{
\alpha\downarrow
\Rightarrow
\text{More smoothing}
}
$$

---

# 27. Simple Exponential Smoothing

**Simple Exponential Smoothing (SES)** is appropriate when the series has approximately:

* Constant level
* No significant trend
* No significant seasonality

The equation is:

$$
\boxed{
SES_t=
\alpha Y_t+
(1-\alpha)SES_{t-1}
}
$$

SES mainly estimates the **level** of the series.

For a basic SES model, forecasts beyond the current period are generally equal to the estimated current level.

---

# 28. Holt's Method

If the series contains a trend, we can use **Holt's method**.

It models:

$$
\boxed{
\text{Level + Trend}
}
$$

The level equation is:

$$
\boxed{
L_t=
\alpha Y_t+
(1-\alpha)(L_{t-1}+T_{t-1})
}
$$

The trend equation is:

$$
\boxed{
T_t=
\beta(L_t-L_{t-1})
+
(1-\beta)T_{t-1}
}
$$

where:

* $L_t$ = estimated level
* $T_t$ = estimated trend
* $\alpha$ = level smoothing parameter
* $\beta$ = trend smoothing parameter

The $h$-step-ahead forecast is:

$$
\boxed{
\hat{Y}_{t+h}=L_t+hT_t
}
$$

---

# 29. Holt-Winters Method

If the series contains:

* Level
* Trend
* Seasonality

we use **Holt-Winters Exponential Smoothing**.

Conceptually:

$$
\boxed{
\text{Holt-Winters}
===================

\text{Level + Trend + Seasonality}
}
$$

For monthly data with yearly seasonality:

$$
\boxed{
s=12
}
$$

Holt-Winters can use either:

* Additive seasonality
* Multiplicative seasonality

---

# 30. Additive vs Multiplicative Smoothing

### Additive Model

In an additive model, the seasonal effect is approximately constant in absolute magnitude.

Conceptually:

$$
\boxed{
Y_t=L_t+T_t+S_t+E_t
}
$$

Use additive seasonality when seasonal fluctuations remain approximately constant over time.

### Multiplicative Model

In a multiplicative model, the seasonal effect changes proportionally with the level.

Conceptually:

$$
\boxed{
Y_t=L_t\times T_t\times S_t\times E_t
}
$$

Use multiplicative seasonality when seasonal fluctuations become larger as the level increases.

For example, a series generated using:

```python
value * seasonal_multiplier
```

represents **multiplicative seasonality**.

# 32. Important Difference: Smoothing vs Forecasting

Smoothing and forecasting are related but not identical.

### Smoothing

The objective is to estimate or visualize the underlying pattern:

$$
\boxed{
\text{Smoothing}
\rightarrow
\text{Reduce Noise}
}
$$

### Forecasting

The objective is to predict future observations:

$$
\boxed{
\text{Forecasting}
\rightarrow
\text{Predict Future}
}
$$

Some smoothing methods can also be used as forecasting methods.

For example:

$$
\boxed{
\text{Trailing MA}
\rightarrow
\text{Smoothing + Forecasting}
}
$$

and:

$$
\boxed{
\text{Exponential Smoothing}
\rightarrow
\text{Smoothing + Forecasting}
}
$$

# 33. Smoothing and Forecast Evaluation

After generating forecasts, evaluate them using metrics such as:

## MAE

$$
\boxed{
MAE=
\frac{1}{n}
\sum_{t=1}^{n}
|Y_t-\hat{Y}_t|
}
$$

## MSE

$$
\boxed{
MSE=
\frac{1}{n}
\sum_{t=1}^{n}
(Y_t-\hat{Y}_t)^2
}
$$

## RMSE

$$
\boxed{
RMSE=
\sqrt{
\frac{1}{n}
\sum_{t=1}^{n}
(Y_t-\hat{Y}_t)^2
}
}
$$

Lower values generally indicate better forecasting performance.
